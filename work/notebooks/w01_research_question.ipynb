## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
from IPython.display import Markdown, display

csv_path = '/home/runner/work/Internship-fly-rank-SEO/Internship-fly-rank-SEO/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(csv_path)

rows = len(df)
clients = df['client_id'].nunique()
decline_share = 100 * df['trend_direction'].eq('down').mean()
zero_click_share = 100 * df['clicks_90d'].eq(0).mean()
median_impr = df['impressions_90d'].median()

quick_look = f"""
### Starter CSV quick look (real numbers)

- Rows: **{rows:,}** across **{clients}** pseudonymized clients.
- Share labeled `down` by `trend_direction`: **{decline_share:.2f}%**.
- Rows with `clicks_90d = 0`: **{zero_click_share:.2f}%**.
- Median `impressions_90d`: **{median_impr:,.0f}**.

These numbers suggest a meaningful prioritization problem: decline is common, and many pages have visibility but weak click outcomes.
"""
display(Markdown(quick_look))
```
